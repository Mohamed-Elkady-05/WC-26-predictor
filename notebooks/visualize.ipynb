```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the simulation results
df = pd.read_csv('../outputs/results/simulation_leaderboard.csv')

# 2. Grab the top 15 teams
top_teams = df.head(15)

# 3. Create the Bar Chart
plt.figure(figsize=(12, 8))
sns.barplot(
    x='Win Probability (%)', 
    y='Country', 
    data=top_teams, 
    palette='mako'
)
```

```python
plt.title('2026 World Cup Win Probability (Monte Carlo Simulation)', fontsize=16, pad=15)
plt.xlabel('Win Probability (%)', fontsize=12)
plt.ylabel('Country', fontsize=12)

# Add the percentage labels to the end of each bar
for index, value in enumerate(top_teams['Win Probability (%)']):
    plt.text(value + 0.2, index, f'{value:.1f}%', va='center')

# 5. Save and show
plt.tight_layout()
plt.savefig('../outputs/results/champions_podium.png', dpi=300)
print("Visualization saved to outputs/results/champions_podium.png")
plt.show()
```
